```python
# Generate Three Input Conditions

LAST_NAME = input("Enter your last name: ").strip().upper()
STUDENT_ID = input("Enter your student ID: ").strip().upper()
FAVORITE_ARTIST = input("Enter your favorite artist: ").strip()

if not LAST_NAME or not FAVORITE_ARTIST:
    raise ValueError("Last name and favorite artist cannot be empty.")

ID_DIGITS = [int(char) for char in STUDENT_ID if char.isdigit()]

if not ID_DIGITS:
    raise ValueError("Student ID must contain at least one digit.")

SEED_NUM = ID_DIGITS[-1]
ID_SUM = sum(ID_DIGITS)
NAME_COUNT = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

BASE_VALUE = ID_SUM + SEED_NUM + NAME_COUNT + ARTIST_LENGTH
DATA_SIZE = 10

ORDERS_SORTED = [
    BASE_VALUE + i * 3 for i in range(DATA_SIZE)
]

ORDERS_REVERSE = ORDERS_SORTED[::-1]

ORDERS_RANDOM = ORDERS_SORTED.copy()
random.Random(ID_SUM + SEED_NUM).shuffle(ORDERS_RANDOM)

DATASETS = {
    "Sorted": ORDERS_SORTED,
    "Reverse-Ordered": ORDERS_REVERSE,
    "Random": ORDERS_RANDOM
}
```

```python
# Implement Three Sorting Algorithms and Record Data

def bubble_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr) - 1):
        swapped = False

        for j in range(len(arr) - 1 - i):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                temp = arr[j]
                arr[j] = arr[j + 1]
                arr[j + 1] = temp
                movements += 3
                swapped = True

        if not swapped:
            break

    return arr, comparisons, movements


def selection_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr) - 1):
        min_index = i

        for j in range(i + 1, len(arr)):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            temp = arr[i]
            arr[i] = arr[min_index]
            arr[min_index] = temp
            movements += 3

    return arr, comparisons, movements


def insertion_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key
        movements += 1

    return arr, comparisons, movements


SORTING_ALGORITHMS = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}
```

```python
# Verify That All Results are Correctly Sorted

results = {}

for condition, orders in DATASETS.items():
    results[condition] = {}

    for algorithm_name, algorithm in SORTING_ALGORITHMS.items():
        start_time = time.perf_counter()

        sorted_orders, comparisons, movements = algorithm(orders)

        execution_time = time.perf_counter() - start_time

        is_sorted = all(
            sorted_orders[i] <= sorted_orders[i + 1]
            for i in range(len(sorted_orders) - 1)
        )

        data_preserved = sorted(orders) == sorted_orders

        results[condition][algorithm_name] = {
            "output": sorted_orders,
            "comparisons": comparisons,
            "movements": movements,
            "time": execution_time,
            "sorted": is_sorted,
            "data_preserved": data_preserved
        }
```

```python
# Compare Performance

comparisons_report = {}

for condition, algorithm_results in results.items():
    minimum_comparisons = min(
        data["comparisons"]
        for data in algorithm_results.values()
    )

    minimum_movements = min(
        data["movements"]
        for data in algorithm_results.values()
    )

    minimum_time = min(
        data["time"]
        for data in algorithm_results.values()
    )

    comparisons_report[condition] = {
        "fewest_comparisons": minimum_comparisons,
        "comparison_winners": [
            name
            for name, data in algorithm_results.items()
            if data["comparisons"] == minimum_comparisons
        ],
        "fewest_movements": minimum_movements,
        "movement_winners": [
            name
            for name, data in algorithm_results.items()
            if data["movements"] == minimum_movements
        ],
        "fastest_time": minimum_time,
        "fastest_algorithms": [
            name
            for name, data in algorithm_results.items()
            if data["time"] == minimum_time
        ]
    }
```

```python
# Relate Results to Big O and Case Analysis

report = []

report.append("=" * 65)
report.append("E-COMMERCE ORDER PROCESSING")
report.append("SORTING ALGORITHM PERFORMANCE REPORT")
report.append("=" * 65)

report.append("\n===INPUT DATASETS===")
for condition, orders in DATASETS.items():
    report.append(f"{condition}: {orders}")

report.append("\n===SORTING RESULTS===")

for condition, algorithm_results in results.items():
    report.append("\n" + "-" * 65)
    report.append(f"CONDITION: {condition}")
    report.append("-" * 65)

    for algorithm_name, data in algorithm_results.items():
        report.append(f"\n{algorithm_name}")
        report.append(f"Sorted Output: {data['output']}")
        report.append(f"Comparisons: {data['comparisons']}")
        report.append(f"Movements: {data['movements']}")
        report.append(f"Execution Time: {data['time']:.8f} seconds")
        report.append(f"Correctly Sorted: {data['sorted']}")
        report.append(f"Data Preserved: {data['data_preserved']}")

    comparison = comparisons_report[condition]

    report.append("\n===PERFORMANCE COMPARISON===")
    report.append(
        f"Fewest Comparisons: {comparison['fewest_comparisons']} "
        f"({', '.join(comparison['comparison_winners'])})"
    )
    report.append(
        f"Fewest Movements: {comparison['fewest_movements']} "
        f"({', '.join(comparison['movement_winners'])})"
    )
    report.append(
        f"Fastest Measured Time: {comparison['fastest_time']:.8f} seconds "
        f"({', '.join(comparison['fastest_algorithms'])})"
    )

report.append("\n===BIG-O ANALYSIS===")

report.append(
    "Bubble Sort: Best O(n), Average O(n^2), Worst O(n^2); "
    "Auxiliary Space O(n) in this implementation."
)
report.append(
    "Selection Sort: Best O(n^2), Average O(n^2), Worst O(n^2); "
    "Auxiliary Space O(n) in this implementation."
)
report.append(
    "Insertion Sort: Best O(n), Average O(n^2), Worst O(n^2); "
    "Auxiliary Space O(n) in this implementation."
)

# Case analysis
report.append("\n===CASE ANALYSIS===")

report.append(
    "Sorted Input: Bubble Sort can stop early, and Insertion Sort "
    "requires few comparisons and movements."
)
report.append(
    "Reverse-Ordered Input: Bubble Sort performs many swaps, while "
    "Insertion Sort performs many shifts. Selection Sort makes a "
    "fixed number of comparisons but typically fewer swaps."
)
report.append(
    "Random Input: Performance depends on the arrangement of values. "
)

print("\n".join(report))
```
