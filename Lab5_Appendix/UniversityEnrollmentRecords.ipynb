```python

# Generate Two Sorted Enrollment Datasets

LAST_NAME = input("Enter your last name: ").strip().upper()
STUDENT_ID = input("Enter your student ID: ").strip().upper()
FAVORITE_ARTIST = input("Enter your favorite artist: ").strip().upper()

ID_DIGITS = [int(char) for char in STUDENT_ID if char.isdigit()]
SEED_NUM = ID_DIGITS[-1] if ID_DIGITS else 0
ID_SUM = sum(ID_DIGITS)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
NAME_COUNT = len(LAST_NAME)

DATASET_A = sorted(
    ID_SUM + SEED_NUM + i * 2
    for i in range(5)
)

DATASET_B = sorted(
    ARTIST_LENGTH + NAME_COUNT + i * 3
    for i in range(5)
)
```

```python
# Count Comparisons

def merge_sorted(a, b):
    merged = []
    i = 0
    j = 0
    comparisons = 0

    while i < len(a) and j < len(b):
        comparisons += 1

        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1

    merged.extend(a[i:])
    merged.extend(b[j:])

    return merged, comparisons
```

```python
# Consolidate Datasets

CONSOLIDATED, COMPARISONS = merge_sorted(DATASET_A, DATASET_B)
```

```python
# Verify Result
ORDERING_VERIFIED = all(
    CONSOLIDATED[i] <= CONSOLIDATED[i + 1]
    for i in range(len(CONSOLIDATED) - 1)
)

RECORD_COUNT_VERIFIED = (
    len(CONSOLIDATED) == len(DATASET_A) + len(DATASET_B)
)
```

```python
# Determine Time and Space Complexity

n = len(DATASET_A)
m = len(DATASET_B)
total_records = len(CONSOLIDATED)

print("\n" + "=" * 50)
print("UNIVERSITY ENROLLMENT RECORDS")
print("=" * 50)

print("\nDepartment A dataset:", DATASET_A)
print("Department B dataset:", DATASET_B)
print("Consolidated dataset:", CONSOLIDATED)

print("\n===ASSESSMENT RESULTS===")
print("Records before consolidation:", len(DATASET_A) + len(DATASET_B))
print("Records after consolidation:", len(CONSOLIDATED))
print("Comparisons performed:", COMPARISONS)
print("Ordering verified:", ORDERING_VERIFIED)
print("Record count verified:", RECORD_COUNT_VERIFIED)

print("\n===COMPLEXITY ANALYSIS===")
print(f"Department A records (n): {n}")
print(f"Department B records (m): {m}")
print(f"Actual comparisons: {COMPARISONS}")
print(f"Time complexity: O(n + m) = O({n} + {m})")
print(f"Auxiliary space complexity: O(n + m) = O({n} + {m})")
print(f"Additional storage: {total_records} records")
```
